# (1) Primary Dataset 1: Yelp Open Dataset

## Dataset Nutrition Label

**Publisher:** Yelp Inc.

**Source:** https://business.yelp.com/data/resources/open-dataset/

**Data Format:** JSON Lines

**Dataset Components:**
- Business metadata: 150,346 records, 14 columns
- Customer reviews: 6,990,280 records, 9 columns

**Research Purpose:** Analyze whether online review characteristics can predict future changes in restaurant popularity, business activity, or closure.

**Unit of Analysis:**
- Original business dataset: one business per record
- Original review dataset: one review per record
- Project analysis: restaurant-month

**Important Identifiers:**
- `business_id`: Links businesses and their reviews
- `review_id`: Identifies individual reviews
- `user_id`: Identifies reviewers

**Temporal Coverage:** February 16, 2005 – January 19, 2022, based on the full review file.

**Geographic Coverage:** Multiple locations across the United States represented in Yelp's dataset.

**Known Limitations:** Self-selection bias in customer reviews, incomplete representation of businesses, and potential differences between online ratings and actual business performance.

## 1. Motivation

The Yelp Open Dataset was created by Yelp Inc. to support academic research and educational projects involving local businesses, customer reviews, and recommendation systems.
The dataset allows researchers to gather consumer preferences, business ratings, review patterns, and relationships between online feedback and business characteristics.
Yelp Inc. created and distributes the dataset. The data originates from activity on Yelp's platform. Specific funding for the dataset release is not identified in the available documentation.
## 2. Composition

The Yelp dataset consists of two primary components:

**Business Metadata**
- Total records: 150,346
- Number of columns: 14
- Unit of observation: One business
- Unique business IDs: 150,346
- Important variables: `business_id`, `name`, `city`, `state`, `stars`, `review_count`, `is_open`, `categories`, and `hours`

**Customer Reviews**
- Total records in the downloaded file: 6,990,280
- Number of columns: 9
- Unit of observation: One customer review
- Unique review IDs: 6,990,280
- Review date range: February 16, 2005 – January 19, 2022
- Important variables: `review_id`, `user_id`, `business_id`, `stars`, `useful`, `funny`, `cool`, `text`, and `date`

The data represents a subset of businesses and reviews available through Yelp rather than every business or customer in the United States.

## 3. Collection Process

Yelp collects business information and customer reviews through its online platform.
Business records contain identifying information, geographic coordinates, categories, operating status, and other descriptive attributes. Reviews contain user-submitted ratings, written feedback, timestamps, and engagement indicators.
Reviews are voluntarily submitted by Yelp users. However, participation on Yelp does not necessarily establish that individual reviewers consented to every possible secondary research use of their content.
The file covers February 2005 through January 2022. The dataset reflects Yelp's selected geographic markets.

## 4. Preprocessing

Yelp provides separate files for business information and reviews, with shared business identifiers that allow the records to be linked.
The publisher has organized the data into structured fields, including numerical ratings, dates, geographic coordinates, and business categories.
The exact internal cleaning, filtering, and moderation procedures used to prepare this release are not fully documented in the files examined, so it is unsure whether what we have is properly given.
The downloaded files serve as our original data source. They are not necessarily the raw, unprocessed records originally stored by Yelp.

## 5. Uses

The Yelp Open Dataset is commonly intended for academic research, including recommendation systems, sentiment analysis, natural language processing, and business-review analysis.
Our intended use is to identify review characteristics associated with later changes in restaurant popularity or operating status.
Ways you shouldnt use it could include review frequency, biased star ratings, and changes in review statemnts week to week.

## 6. Distribution

The dataset is published by Yelp Inc. through its Open Dataset program.
**Official source:** https://business.yelp.com/data/resources/open-dataset/
**File format:** JSON Lines
**Access:** Downloadable through Yelp's dataset distribution process.
**Terms:** Yelp provides the dataset under its dataset agreement for permitted academic and educational uses. It should not be treated as unrestricted public-domain data.
Any redistribution, publication of review text, or commercial use must comply with the applicable Yelp dataset agreement.

## 7. Maintenance
Yelp Inc. maintains the Yelp platform and controls the distribution of the Open Dataset.
The downloaded files represent a historical snapshot, with the latest review dated January 19, 2022 so its not as up to date as we would like.
Although Yelp's live platform continues to receive reviews and business updates, those changes do not automatically appear in our local downloaded files.
To identify dataset updates, researchers should check Yelp's official dataset page, release information, and available file metadata.

# (1) Primary Dataset 2: Google Local Reviews (2021)

## Dataset Nutrition Label

**Publisher:** UC San Diego (UCSD), McAuley Lab

**Source:** https://mcauleylab.ucsd.edu/public_datasets/gdrive/googlelocal/

**Original Platform:** Google Maps

**Format:** Compressed JSON Lines (`.json.gz`)

**Geographic Coverage:** Illinois, United States

**Business Metadata:** 179,205 records, 15 columns

**Customer Reviews:** 23,096,838 records, 8 columns

**Restaurant Subset:** 28,809 business records identified in the previous project checkpoint

**Review Date Range:** December 31, 1990 – September 8, 2021, according to the downloaded file. The unusually early date requires validation.

**Linking Variable:** `gmap_id`

**Research Purpose:** Study whether changes in customer ratings, review frequency, and review sentiment are associated with subsequent restaurant popularity or closure.

## 1. Motivation

The Google Local Reviews dataset was assembled by researchers at UC San Diego to support academic research involving recommendation systems, business information, customer reviews, and geographic data.
The dataset contains information originating from Google Maps, including business locations, categories, ratings, written reviews, and user interactions.
The 2021 release identifies Tianyang Zhang and Jiacheng Li at UCSD as contributors. Specific external funding for the dataset collection is not established by the documentation reviewed.
Our project uses the Illinois portion to examine relationships between online review patterns and restaurant outcomes over time.

## 2. Composition

**Business Metadata**
- 179,205 records
- 15 columns
- One record represents a business listing
- 178,204 unique non-null `gmap_id` values
- Variables include business name, address, coordinates, category, average rating, review count, price, hours, and business status

The difference between total rows and unique identifiers indicates possible repeated or missing IDs requiring further investigation.
The previous checkpoint identified 28,809 restaurant records using business category information.

**Customer Reviews**
- 23,096,838 records
- 8 columns
- One record represents a customer review
- Variables include reviewer identifier, reviewer name, timestamp, rating, review text, pictures, business response, and business identifier

Reviews are linked to businesses through `gmap_id`.
The dataset represents businesses and reviewers visible in the collected Google Maps data, not every business or customer in Illinois.

## 3. Collection Process

The dataset was compiled from Google Maps business listings and user reviews by UCSD researchers.
The collection includes business metadata and user-generated content such as star ratings, review text, and images.
The published 2021 dataset contains information collected through September 2021.
Review timestamps allow the project to examine changes in review activity over time. However, the earliest timestamp found in our downloaded file is December 31, 1990, which may indicate a data-quality problem and should be investigated before temporal modeling.
Individual consent for inclusion in this research dataset has not been released so we dont know.

## 4. Preprocessing

The UCSD release organizes Google Maps information into structured business metadata and review records.
Business attributes include numerical ratings, coordinates, categories, and descriptive fields. Review records contain timestamps, ratings, text, and identifiers.
The downloaded JSON files are the original files used in our analysis, although were not sure if they reflect processing by the dataset publisher.

## 5. Uses

The Google Local Reviews dataset supports academic research in recommendation systems, review analysis, geographic business analysis, and consumer feedback.
Our intended use is to analyze whether earlier review characteristics are associated with later changes in restaurant popularity or operating status.
Ways you should not use it could include misrepresentation of data through taking a group of people as a whole section and not individually.

## 6. Distribution

The dataset is distributed through UC San Diego's public research dataset repository.
**Official source:** https://mcauleylab.ucsd.edu/public_datasets/gdrive/googlelocal/
The Illinois files are distributed as compressed JSON Lines files.
The files are publicly accessible for research. However, a specific unrestricted reuse license is not clearly identified in the dataset documentation review.

# (6) Ethical Considerations and Representation

## Representation of People

Both Yelp and Google Local Reviews contain information generated by individuals who voluntarily submit online business reviews.
Yelp provides reviewer identifiers and written reviews. Google Local Reviews includes reviewer identifiers, names, ratings, text, and potentially photographs.
These records represent individuals who use the platforms and choose to leave reviews. They do not constitute a random or representative sample of all restaurant customers.
People who do not use online review platforms, have limited internet access, or are less comfortable writing public reviews may be underrepresented.

## Potential Biases

Customers who leave reviews may differ systematically from those who do not.
Yelp includes selected geographic markets, while the Google dataset used here covers Illinois. Findings may not generalize to other locations.
Popular businesses may receive more reviews, making them more visible in the dataset than smaller or less frequently reviewed establishments.
Customers with unusually positive or negative experiences may be more motivated to post reviews.
Both datasets are historical snapshots. Changes in consumer behavior, business conditions, and review platforms may limit their relevance to current conditions.

## CARE Principles

**Collective Benefit:** The research should contribute to understanding business trends and help identify useful patterns in customer feedback without unfairly disadvantaging particular communities.

**Authority to Control:** Data subjects and communities may have interests in how information about them is used.

**Responsibility:** Researchers should document data limitations, evaluate bias, and avoid making unsupported claims about individual restaurants or neighborhoods.

**Ethics:** The analysis should protect reviewer privacy, acknowledge incomplete representation, and avoid conclusions that unfairly stigmatize businesses or communities.

## People Affected by the Analysis

Restaurant owners, employees, customers, and surrounding communities may be affected by conclusions drawn from these datasets.
Such predictions could damage a business's reputation if presented without appropriate uncertainty and validation.
