```python
import pandas as pd
import numpy as numpy
import json
import gzip

pd.set_option("display.max_columns", None)
```

```python
# Chicago Food Inspections

food_url = (
    "https://data.cityofchicago.org/resource/"
    "4ijn-s7e5.csv?$limit=50000"
)

food_inspections = pd.read_csv(food_url)

print("Shape:", food_inspections.shape)

display(food_inspections.head())

print("\nColumn types:")
print(food_inspections.dtypes)
```

```python
food_columns = [
    "dba_name",
    "license_",
    "facility_type",
    "address",
    "inspection_date",
    "results",
    "violations",
    "latitude",
    "longitude"
]

display(food_inspections[food_columns].head())
```

Chicago Business Licenses

```python
# Chicago Business Licenses

license_url = (
    "https://data.cityofchicago.org/resource/"
    "r5kz-chrr.csv?$limit=50000"
)

business_licenses = pd.read_csv(
    license_url,
    low_memory=False
)

print("Shape:", business_licenses.shape)

display(business_licenses.head())

print("\nColumn types:")
print(business_licenses.dtypes)
```

```python
business_licenses.columns.tolist()
```

```python
def summarize_dataset(df, name):
    print("=" * 60)
    print(name)
    print("=" * 60)

    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])

    print("\nColumn names:")
    print(df.columns.tolist())

    print("\nData types:")
    print(df.dtypes)

    print("\nFirst 5 rows:")
    display(df.head())
```

```python
summarize_dataset(
    food_inspections,
    "Chicago Food Inspections"
)
```

```python
summarize_dataset(
    business_licenses,
    "Chicago Business Licenses"
)
```

Google Local Reviews

```python
# Google local reviews
google_metadata = pd.read_json(
    "data/meta-Illinois.json.gz",
    lines=True,
    compression="gzip"
)

print("Shape:", google_metadata.shape)

display(google_metadata.head())

print(google_metadata.dtypes)
```

```python
google_metadata["category"].head()
```

```python
google_restaurants = google_metadata[
    google_metadata["category"].apply(
        lambda x: isinstance(x, list)
        and any(
            "restaurant" in str(category).lower()
            for category in x
        )
    )
].copy()

print("All Illinois businesses:", google_metadata.shape)
print("Illinois restaurants:", google_restaurants.shape)

display(google_restaurants.head())
```

```python
google_reviews = pd.read_json(
    "data/review-Illinois.json.gz",
    lines=True,
    compression="gzip",
    nrows=100000
)

print("Shape:", google_reviews.shape)

display(google_reviews.head())

print(google_reviews.dtypes)
```

```python
google_review_columns = [
    "gmap_id",
    "rating",
    "text",
    "time",
    "resp"
]

display(google_reviews[google_review_columns].head())
```
